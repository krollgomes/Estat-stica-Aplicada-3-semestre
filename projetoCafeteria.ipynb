```python
print("\n--- Relação Preço vs Vendas (com Promoção) ---")

plt.figure(figsize=(10, 6))
sns.scatterplot(data=dados,
                x='Preco_Cafe',
                y='Vendas_Cafe',
                hue='Promocao',
                style='Promocao',
                s=100
            )
plt.xlabel('Preço')
plt.ylabel('Quantidade vendida')
plt.title('Relação entre o preço e as vendas do café')
plt.grid(True)
plt.show()
```

```python
# Calcula a média
media = dados['Vendas_Cafe'].mean()

# Cria a variável categórica
dados['variavel'] = np.where(
    dados['Vendas_Cafe'] > media,
    'Acima_da_media',
    'Abaixo_da_media'
)

# Exibe contagem
print("\nContagem de Vendas Acima/Abaixo da Média:")
contagem = dados['variavel'].value_counts()
print(contagem)

# Gráfico de barras
contagem.plot(
    kind='bar',
    color=['green', 'red'],
    edgecolor='black'
)

plt.title('Contagem Acima vs. Abaixo da Média')
plt.ylabel('Contagem')
plt.xticks(rotation=0)

plt.show()
```

```python
print("\n--- Gerando Boxplot: Vendas por Status da Promoção ---")

plt.figure(figsize=(8, 7))

sns.boxplot(
    data=dados,
    x='Promocao',
    y='Vendas_Cafe',
    palette=["#ff6347", "#90ee90"]
)

plt.title('Comparação de Vendas: Com vs. Sem Promoção', fontsize=16)
plt.xlabel('Promoção Ativa?', fontsize=12)
plt.ylabel('Vendas (Quantidade)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
formula = 'Vendas_Cafe ~ Preco_Cafe'

modelo = ols(formula, data=dados)

resultados = modelo.fit()

print(resultados.summary())

print("\n--- Gerando gráfico de regressão (Dispersão + Linha)...")

plt.figure(figsize=(10, 6))

sns.regplot(
    data=dados,
    x='Preco_Cafe',
    y='Vendas_Cafe',
    scatter_kws={'color': 'blue'},
    line_kws={'color': 'red'}
)

plt.title('Relação entre o Preço e as Vendas do Café', fontsize=16)
plt.xlabel('Preço do Café', fontsize=12)
plt.ylabel('Vendas de Café', fontsize=12)
plt.grid(True)

plt.show()
```

```python
dados['Promocao'] = dados['Promocao'].astype('category')
formula_multipla = 'Vendas_Cafe ~ Preco_Cafe + Promocao + Preco_Leite'
modelo_multiplo = ols(formula=formula_multipla, data=dados)
resultados_multiplos = modelo_multiplo.fit()
print("=== Resultados da Regressão Múltipla ===")
print(resultados_multiplos.summary())
```

```python
intercepto = resultados_multiplos.params['Intercept']
coef_preco_cafe = resultados_multiplos.params['Preco_Cafe']
coef_preco_leite = resultados_multiplos.params['Preco_Leite']

x_surf = np.linspace(dados['Preco_Cafe'].min(), dados['Preco_Cafe'].max(), 10)
y_surf = np.linspace(dados['Preco_Leite'].min(), dados['Preco_Leite'].max(), 10)
x_surf, y_surf = np.meshgrid(x_surf, y_surf)

z_surf = intercepto + (coef_preco_cafe * x_surf) + (coef_preco_leite * y_surf)

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    dados['Preco_Cafe'],
    dados['Preco_Leite'],
    dados['Vendas_Cafe'],
    c='blue',
    marker='o'
)

ax.plot_surface(x_surf, y_surf, z_surf, color='red', alpha=0.4)

ax.set_xlabel('Preço do Café (X1)')
ax.set_ylabel('Preço do Leite (X2)')
ax.set_zlabel('Vendas de Café (Y)')
ax.set_title('Regressão Múltipla 3D')

plt.show()
```

```python
dados = pd.DataFrame(dados_dict)

dados['Promocao'] = dados['Promocao'].astype('category')

formula_completa = 'Vendas_Cafe ~ Preco_Cafe + Promocao + Preco_Leite'
modelo_completo = ols(formula=formula_completa, data=dados).fit()

print("=== Resultados do Modelo Completo (Com Promoção) ===")
print(modelo_completo.summary())

formula_sem_promocao = 'Vendas_Cafe ~ Preco_Cafe + Preco_Leite'
modelo_sem_promocao = ols(formula=formula_sem_promocao, data=dados).fit()

print("\n=== Resultados do Modelo Reduzido (Sem Promoção) ===")
print(modelo_sem_promocao.summary())
```

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt

anova = sm.stats.anova_lm(modelo_sem_promocao, modelo_completo)
print(anova)

fig = sm.graphics.plot_partregress_grid(modelo_completo)
fig.tight_layout()
plt.show()
```


## Análise dos Resultados

### O preço do café influencia a venda?

Sim. A regressão linear simples mostrou uma relação negativa entre o preço do café e as vendas. Isso significa que, conforme o preço aumenta, a quantidade vendida tende a diminuir.

O gráfico de dispersão com linha de regressão evidencia uma tendência decrescente, indicando comportamento compatível com a lei da demanda.

---

### Promoções fazem diferença?

Sim. O boxplot comparando períodos com e sem promoção mostra que as vendas tendem a ser maiores quando há promoção.

Além disso, no modelo de regressão múltipla, a variável `Promocao` contribui positivamente para explicar as vendas, indicando que campanhas promocionais aumentam a quantidade vendida.

---

### Quantas vezes as vendas ficaram acima/abaixo da média?

A variável categórica criada no script separa as vendas em:

- `Acima_da_media`
- `Abaixo_da_media`

Como existem 30 observações, a contagem ficou aproximadamente equilibrada entre os dois grupos, permitindo comparar períodos de desempenho superior e inferior à média geral de vendas.

---

### Como é a relação entre a variável independente e as vendas? É crescente, decrescente ou curva?

A relação entre `Preco_Cafe` e `Vendas_Cafe` é predominantemente decrescente e aproximadamente linear.

Os gráficos de dispersão indicam que preços menores estão associados a maiores volumes de vendas. Não há evidências fortes de comportamento curvilíneo nos dados analisados.

---

### Há uma faixa da variável independente onde a previsão de vendas é menos confiável?

Sim. As previsões tendem a ser menos confiáveis nas extremidades da variável `Preco_Cafe`, especialmente para preços muito baixos ou muito altos.

Isso ocorre porque existem menos observações nessas regiões, reduzindo a capacidade do modelo de generalizar corretamente. A maior concentração de dados ocorre em torno da faixa intermediária de preços.

---

### Qual variável tem maior efeito nas vendas?

O `Preco_Cafe` aparenta ser a variável com maior impacto nas vendas, devido ao coeficiente negativo mais relevante no modelo.

A variável `Promocao` também exerce influência importante, aumentando as vendas quando ativa.

Já `Preco_Leite` possui efeito menor comparado ao preço do café e à promoção.

---

### Análise do erro residual e do R². Os valores são bons ou ruins? Compare o resultado da regressão simples com a múltipla. Qual o melhor modelo?

O modelo de regressão múltipla apresentou melhor desempenho em relação ao modelo simples.

Principais pontos da comparação:

- O R² do modelo múltiplo é maior, indicando maior capacidade de explicar a variação das vendas.
- Os resíduos ficaram mais distribuídos e com menor erro no modelo múltiplo.
- A inclusão das variáveis `Promocao` e `Preco_Leite` aumentou o poder explicativo do modelo.

A ANOVA entre os modelos também ajuda a verificar se a inclusão da variável `Promocao` melhora significativamente o ajuste.

Assim, o modelo múltiplo é o mais adequado, pois considera mais fatores que influenciam as vendas e apresenta melhor qualidade estatística.
